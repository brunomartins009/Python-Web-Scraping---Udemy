### Montando uma string com o comando de insert

```python
inserir_contato = ("insert into contatos (nome, telefone, celular)"
                    "values ('Pateta', '(27)1111-1111', '(27)12345-3333')")
# Executando o comando insert
cursor.execute(inserir_contato)
# Gravando a informação permanentemente
conexao.commit()
# Fechando o cursor
cursor.close()
# Fechando a conexão
conexao.close()
```

### Usando o comando `update` para alterar um registro

```python
conexao_update = mysql.connector.connect(user='root', password='@bmc009@', host='127.0.0.1', database='agenda')
cursor_update = conexao_update.cursor()

atualizar_contato = ("update contatos set nome = %s, telefone = %s, celular = %s where id = 1")
contato_update = ('Pocahontas', '(99)9282-1827', '(28)09876-2728')

cursor_update.execute(atualizar_contato, contato_update)
conexao_update.commit()
cursor_update.close()
conexao_update.close()
```

### Apagando um registro com o comando `delete`
